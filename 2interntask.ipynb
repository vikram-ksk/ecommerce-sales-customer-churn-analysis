### cohort Analysis

```python
df["Purchase_Month"] = df["Order Date"].dt.to_period("M")
```

```python
df["Cohort_Month"] = (
    df.groupby("Customer ID")["Purchase_Month"]
      .transform("min")
)
```

```python
df["Cohort_Index"] = (
    (df["Purchase_Month"].dt.year - df["Cohort_Month"].dt.year) * 12
    + (df["Purchase_Month"].dt.month - df["Cohort_Month"].dt.month)
    + 1
)
```

```python
cohort_data = (
    df.groupby(["Cohort_Month", "Cohort_Index"])["Customer ID"]
      .nunique()
      .reset_index()
)

cohort_data.head(20)
```

```python
cohort_pivot = cohort_data.pivot(
    index="Cohort_Month",
    columns="Cohort_Index",
    values="Customer ID"
)

cohort_retention = cohort_pivot.div(
    cohort_pivot[1],
    axis=0
) * 100

cohort_retention.round(2)
cohort_retention.round(2).head(10)
```

### RFM SEGMENTATION

```python
analysis_date = df["Order Date"].max() + pd.Timedelta(days=1)

print("Analysis Date:", analysis_date)
```

```python
rfm = df.groupby("Customer ID").agg({
    "Order Date": lambda x: (analysis_date - x.max()).days,
    "Order ID": "nunique",
    "Sales": "sum"
}).reset_index()

rfm.columns = ["Customer ID", "Recency", "Frequency", "Monetary"]

rfm.head()
```

```python
rfm.describe()
```

```python
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(int).astype(str) +
    rfm["F_Score"].astype(int).astype(str) +
    rfm["M_Score"].astype(int).astype(str)
)

rfm.head(10)
```

```python
def rfm_segment(row):
    r = row["R_Score"]
    f = row["F_Score"]
    m = row["M_Score"]

    if r >= 4 and f >= 4 and m >= 4:
        return "Champions"

    elif r >= 3 and f >= 3:
        return "Loyal Customers"

    elif r >= 4 and f <= 2:
        return "New Customers"

    elif r <= 2 and f >= 3:
        return "At Risk"

    elif r <= 2 and f <= 2 and m <= 2:
        return "Lost"

    else:
        return "Potential Loyalists"


rfm["Segment"] = rfm.apply(rfm_segment, axis=1)

rfm.head(10)
```

```python
segment_summary = (
    rfm["Segment"]
    .value_counts()
    .reset_index()
)

segment_summary.columns = ["Segment", "Customer_Count"]

segment_summary
```

```python
segment_revenue = (
    rfm.groupby("Segment")["Monetary"]
       .sum()
       .sort_values(ascending=False)
       .reset_index()
)

segment_revenue
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("Libraries imported successfully!")
```

```python
cutoff = pd.Timestamp("2017-10-01")

future_end = cutoff + pd.Timedelta(days=90)

print("Historical cutoff:", cutoff.date())
print("Future 90-day end:", future_end.date())
```

```python
# Data before the cutoff
hist = df[df["Order Date"] < cutoff].copy()

# Data in the following 90 days
future = df[
    (df["Order Date"] >= cutoff) &
    (df["Order Date"] <= future_end)
].copy()

# Customer-level historical features
customer_features = hist.groupby("Customer ID").agg(
    Recency=("Order Date", lambda x: (cutoff - x.max()).days),
    Frequency=("Order ID", "nunique"),
    Monetary=("Sales", "sum"),
    Avg_Order_Value=("Sales", "mean"),
    Total_Quantity=("Quantity", "sum")
).reset_index()

print(customer_features.head())
print("Total customers:", len(customer_features))
```

```python
future_orders = future.groupby("Customer ID")["Order ID"].nunique()

customer_features["Future_Orders_90D"] = (
    customer_features["Customer ID"]
    .map(future_orders)
    .fillna(0)
)

print(customer_features[
    ["Customer ID", "Future_Orders_90D"]
].head(10))
```

```python
customer_features["Churn"] = (
    customer_features["Future_Orders_90D"] == 0
).astype(int)

print(customer_features["Churn"].value_counts())
```

```python
X = customer_features[
    [
        "Recency",
        "Frequency",
        "Monetary",
        "Avg_Order_Value",
        "Total_Quantity"
    ]
]

y = customer_features["Churn"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
model = DecisionTreeClassifier(
    max_depth=4,
    class_weight="balanced",
    random_state=42
)

model.fit(X_train, y_train)

print("Decision Tree model trained successfully!")
```

```python
y_pred = model.predict(X_test)

print(y_pred[:20])
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

print("Accuracy :", round(accuracy * 100, 2), "%")
print("Precision:", round(precision * 100, 2), "%")
print("Recall   :", round(recall * 100, 2), "%")
print("F1 Score :", round(f1 * 100, 2), "%")
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```
